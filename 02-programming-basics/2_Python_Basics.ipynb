## Exercise 2: Writing a Moving Average Function

```python
def moving_average(prices, window):
   ma_values = []

   for i in range(window, len(prices) + 1):
       window_prices = prices[i - window:i]
       average = sum(window_prices) / window
       ma_values.append(average)

   return ma_values


prices = [100, 102, 101, 105, 110]
ma_3 = moving_average(prices, 3)

print(ma_3)
```

## Exercise 4: Portfolio Value Calculation

```python
portfolio = {
   "AAPL": 10,
   "TSLA": 5
}

prices = {
   "AAPL": 150,
   "TSLA": 700
}

total_value = 0

for symbol in portfolio:
   shares = portfolio[symbol]
   price = prices[symbol]
   total_value += shares * price

print("Total portfolio value:", total_value)
```

## Exercise 5: Maximum Drawdown (Conceptual Intro)

```python
prices = [100, 120, 115, 130, 90, 95]

max_price = prices[0]
max_drawdown = 0

for price in prices:
    if price > max_price:
        max_price = price

    drawdown = (price - max_price) / max_price
    if drawdown < max_drawdown:
        max_drawdown = drawdown
print("Maximum drawdown:", max_drawdown)
```

## Exercise 7: Counting Winning and Losing Days

```python
returns = [0.02, -0.01, 0.03, -0.04, 0.01]

wins = 0
losses = 0

for r in returns:
    if r > 0:
        wins += 1
    elif r < 0:
        losses += 1

print("Winning days:", wins)
print("Losing days:", losses)
```
