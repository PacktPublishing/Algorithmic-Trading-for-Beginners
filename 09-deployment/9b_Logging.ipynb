```python
import logging

# Clear existing handlers to allow reconfiguration
# Not needed in plain text Python file
root = logging.getLogger()
if root.hasHandlers():
    for handler in root.handlers[:]:
        root.removeHandler(handler)

logging.basicConfig(
    filename="trading.log",
    level=logging.INFO,
    format="%(asctime)s %(levelname)s %(message)s"
)
```

```python
logging.info("Strategy started")
logging.info("Signal generated: long")
logging.warning("Insufficient funds")
logging.error("Order submission failed")
logging.shutdown()
```
