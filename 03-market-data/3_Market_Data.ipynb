# Vectorizing DataFrame operations

## Computing simple returns

```python
data["Return"] = data["Close"].pct_change()

data.head()
```

## Creating lagged variables

```python
data["Return_lag1"] = data["Return"].shift(1)
data.head()
```

## Computing log prices

```python
import numpy as np

data["LogPrice"] = np.log(data["Close"])
data.head()
```

## Computing log returns

```python
# OK method
data["LogReturn"] = np.log(data["Close"] / data["Close"].shift(1))
data.head()
```

```python
# Better method
data["LogReturn"] = np.log(data["Close"]).diff()
data.head()
```

## Rolling Statistics

```python
# 20-day rolling mean
data["MA20"] = data["Close"].rolling(20).mean()
```

```python
# 20-day rolling volatility (standard deviation)
data["Volatility20"] = data["Return"].rolling(20).std()
```

# Plotting financial data

```python
import matplotlib.pyplot as plt
```

## Plotting a Price Series

```python
data["Close"].plot(figsize=(10,5))
plt.title("Stock price over time")
plt.xlabel("Date")
plt.ylabel("Price")
plt.show()
```

## Plotting Log Prices

```python
np.log(data["Close"]).plot(figsize=(10,5))
plt.title("Log price over time")
plt.ylabel("Log Price")
plt.show()
```

## Histogram of Returns

```python
data["Return"].hist(bins=50)
plt.title("Distribution of daily returns")
plt.xlabel("Return")
plt.ylabel("Frequency")
plt.show()
```

## Scatterplot of Consecutive Returns

```python
data["Return_lag1"] = data["Return"].shift(1)
plt.scatter(data["Return_lag1"], data["Return"], alpha=0.5)
plt.xlabel("Yesterday's Return")
plt.ylabel("Today's Return")
plt.title("Return vs previous day return")
plt.show()
```

## Plotting Moving Averages

```python
data["MA20"] = data["Close"].rolling(20).mean()
data["MA50"] = data["Close"].rolling(50).mean()

data[["Close","MA20","MA50"]].plot(figsize=(10,5))
plt.title("Price with moving averages")
plt.show()
```

## Rolling Volatility

```python
data["Volatility20"] = data["Return"].rolling(20).std()
data["Volatility20"].plot(figsize=(10,5))
plt.title("20-day rolling volatility")
plt.show()
```
