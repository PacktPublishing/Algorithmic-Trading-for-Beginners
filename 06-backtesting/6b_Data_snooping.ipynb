```python
import numpy as np
from scipy.stats import pearsonr

np.random.seed(123)

N = 100

# Random target
y = np.random.randn(N)

best_corr = 0
best_p = None
best_feature = None

for i in range(1000):
    x = np.random.randn(N)

    corr, p = pearsonr(x, y)

    if abs(corr) > abs(best_corr):
        best_corr = corr
        best_p = p
        best_feature = i

print(f"Best feature: {best_feature}")
print(f"Correlation: {best_corr:.3f}")
print(f"P-value: {best_p:.6f}")
```
